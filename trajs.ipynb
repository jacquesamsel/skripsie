```python
import torch
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
import os
from tqdm import tqdm
```

```python
trajectories = []
for traj in tqdm(os.listdir("results/trajectories")):
    traj_vals = torch.load(f"results/trajectories/{traj}", weights_only=False)
    trajectories.append(traj_vals)
```

```python
pos = 0
plt.plot(trajectories[pos]["agents", "x"].reshape(-1), trajectories[pos]["agents", "z"].reshape(-1))
```
