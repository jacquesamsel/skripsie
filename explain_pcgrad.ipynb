```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
xs = np.linspace(-4, 2, 1000)
ys = np.linspace(-4, 6, 1000)

X, Y = np.meshgrid(xs, ys)

primary_fn = lambda x, y: (x + 2 * np.tanh(y - 0.5)) ** 2
Z_primary = primary_fn(X, Y)

aux_fn = lambda x, y: (x + 2 * np.tanh(y * 2)) ** 2 + 0.05 * y**2
Z_aux = aux_fn(X, Y)

Z = Z_primary + Z_aux

plt.contourf(X, Y,
    Z,
    levels=np.geomspace(Z.min() + 1e-3, Z.max(), 100),
    cmap="Oranges_r"
)
```
