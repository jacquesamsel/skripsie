```python
import pandas as pd
import seaborn as sns
import json
```

```python
data = pd.read_csv("aux_head_true_values.csv")
```

```python
data.describe()
```

```python
sns.boxplot(data=data).set_xticklabels(labels=data.columns, rotation=90)
```

```python
means = data.mean()
var = data.var()
median = data.median()
with open("results/aux_head_scaling_factors.json", "w") as f:
    json.dump({
        key: {
            "mean": means[key],
            "var": var[key],
            "mode": median[key]
        }
        for key in data.columns
    }, f)
```
